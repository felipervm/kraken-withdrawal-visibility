```python
from pathlib import Path
import sys
import pandas as pd
root = Path.cwd()
if not (root / 'data/synthetic_review_cases.csv').exists():
    root = root.parent
sys.path.insert(0, str(root / 'analysis'))
from validate import validate, verify_generator
verify_generator()
validation = validate(root / 'data/synthetic_review_cases.csv')
print(validation)
df = pd.read_csv(root / 'data/synthetic_review_cases.csv')
```

```python
qa = ['status_stale','missing_support_reference','cross_channel_mismatch','missed_permitted_update']
assert df[qa].isin([0,1]).all().all()
assert (df[qa].max(axis=1) == df.communication_exception).all()
summary = df.groupby('review_type').agg(cases=('case_id','size'), flagged=('communication_exception','sum'), avg_repeat=('repeat_contacts_7d','mean'))
summary['flag_rate'] = summary.flagged / summary.cases
display(summary)
```

## Interpretation
These output values are **mechanical consequences of arbitrary injected conditions**. They are not evidence of real Kraken exception rates, customer confusion, effects on support, or return on investment. Real evaluation requires authorized data access, stakeholder approval and an appropriate baseline.


## Reproducibility and scope
The original CSV is a preserved input; its original generation code was not supplied. `generate_fixture.py` creates a separate deterministic fixture with matching arbitrary case and exception totals, not the original rows. `validate.py` cross-checks the CSV with the supplied SQL using SQLite.
The four flags are pre-labeled simulation inputs. This notebook validates their domain, union and aggregation; it does not detect actual message staleness or missing updates from production events. An internal audit needs timestamped approved states, observed channel messages, delivery records and agreed eligibility rules.
